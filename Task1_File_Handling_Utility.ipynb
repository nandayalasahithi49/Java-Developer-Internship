```python
%%writefile FileHandlingUtility.java
import java.io.*;

public class FileHandlingUtility {

    public static void main(String[] args) {

        try {

            FileWriter writer = new FileWriter("sample.txt");

            writer.write("Hello Java\n");
            writer.write("Internship Task 1\n");
            writer.close();

            System.out.println("Data Written");

            BufferedReader reader =
                    new BufferedReader(new FileReader("sample.txt"));

            String line;

            System.out.println("\nReading File:");

            while ((line = reader.readLine()) != null) {
                System.out.println(line);
            }

            reader.close();

            FileWriter appendWriter =
                    new FileWriter("sample.txt", true);

            appendWriter.write("New Line Added\n");
            appendWriter.close();

            System.out.println("\nFile Modified");

        } catch (Exception e) {
            System.out.println(e);
        }
    }
}

```
